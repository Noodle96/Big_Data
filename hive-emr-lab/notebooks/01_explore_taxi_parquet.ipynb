## Comparacion de columnas entre 2023 y 2024 de Yellow Taxi Trip Record 

```python
from pathlib import Path
import pandas as pd

file_2023: Path = Path("../data/taxi/raw/yellow_tripdata_2023-01.parquet")
file_2024: Path = Path("../data/taxi/raw/yellow_tripdata_2024-01.parquet")

df_2023 = pd.read_parquet(file_2023, engine="pyarrow")
df_2024 = pd.read_parquet(file_2024, engine="pyarrow")

cols_2023 = set(df_2023.columns)
cols_2024 = set(df_2024.columns)

print("Solo en 2023:")
print(sorted(cols_2023 - cols_2024))

print("\nSolo en 2024:")
print(sorted(cols_2024 - cols_2023))

print("\nMismas columnas:")
print(cols_2023 == cols_2024)
```

```python
print(df_2023["airport_fee"].dtype)

print(df_2024["Airport_fee"].dtype)
```

```python
print(df_2023.dtypes)

print(df_2024.dtypes)
```

```python
for col in sorted(set(df_2023.columns) & set(df_2024.columns)):
    if df_2023[col].dtype != df_2024[col].dtype:
        print(
            col,
            df_2023[col].dtype,
            df_2024[col].dtype
        )
```
