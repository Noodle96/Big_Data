```python
from pathlib import Path
from typing import List

import pandas as pd
```

```python
raw_csv_path: Path = Path("../data/official_onpe/raw/PR-ESP_Presidencial_2026-05-11_03-15-PM_99.668.csv")

df: pd.DataFrame = pd.read_csv(raw_csv_path)
```

```python
metadata_cols: List[str] = [
    "TIPO DE ELECCIÓN",
    "ÁMBITO",
    "REGIÓN / CONTINENTE",
    "PROVINCIA / PAÍS",
    "DISTRITO / CIUDAD",
    "CENTRO POBLADO",
    "LOCAL DE VOTACIÓN",
    "NÚMERO DE MESA",
    "ESTADO DEL ACTA",
    "ELECTORES HÁBILES",
]

vote_cols: List[str] = [col for col in df.columns if col not in metadata_cols]

df["TOTAL_VOTOS_EMITIDOS"] = df[vote_cols].sum(axis=1)

participacion_input_df: pd.DataFrame = df[
    [
        "REGIÓN / CONTINENTE",
        "NÚMERO DE MESA",
        "ELECTORES HÁBILES",
        "TOTAL_VOTOS_EMITIDOS",
    ]
].copy()

processed_dir: Path = Path("../data/official_onpe/processed")
processed_dir.mkdir(parents=True, exist_ok=True)

output_path: Path = processed_dir / "participacion_region_input.tsv"

participacion_input_df.to_csv(output_path, index=False, sep="\t")

output_path
```
