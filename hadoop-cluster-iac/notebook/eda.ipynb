```python
from pathlib import Path
from typing import List

import pandas as pd
import matplotlib.pyplot as plt
```

```python
csv_path: Path = Path(
    "../data/official_onpe/raw/PR-ESP_Presidencial_2026-05-11_03-15-PM_99.668.csv"
)

df: pd.DataFrame = pd.read_csv(csv_path)

df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
df.columns.tolist()
```

```python
metadata_cols: List[str] = [
    "TIPO DE ELECCIÓN",
    "ÁMBITO",
    "REGIÓN / CONTINENTE",
    "PROVINCIA / PAÍS",
    "DISTRITO / CIUDAD",
    "CENTRO POBLADO",
    "LOCAL DE VOTACIÓN",
    "NÚMERO DE MESA",
    "ESTADO DEL ACTA",
    "ELECTORES HÁBILES",
]

vote_cols: List[str] = [col for col in df.columns if col not in metadata_cols]

print("Columnas metadata:", len(metadata_cols))
print("Columnas votos:", len(vote_cols))
vote_cols
```

```python
df.describe()
```

```python
numeric_cols: List[str] = ["ELECTORES HÁBILES"] + vote_cols

df[numeric_cols].describe()
```

```python
missing_values: pd.Series = df.isna().sum().sort_values(ascending=False)

missing_values[missing_values > 0]
```

```python
df["ESTADO DEL ACTA"].value_counts()
```

```python
estado_counts: pd.Series = df["ESTADO DEL ACTA"].value_counts()
print(estado_counts)

plt.figure(figsize=(8, 4))
estado_counts.plot(kind="bar")
plt.title("Cantidad de mesas por estado del acta")
plt.xlabel("Estado del acta")
plt.ylabel("Cantidad de mesas")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
region_counts: pd.Series = df["REGIÓN / CONTINENTE"].value_counts()

region_counts.head(50)
```

```python
total_votes_by_party: pd.Series = df[vote_cols].sum().sort_values(ascending=False)

total_votes_by_party
```

```python
plt.figure(figsize=(12, 6))
total_votes_by_party.head(15).plot(kind="bar")
plt.title("Top 15 organizaciones por votos totales")
plt.xlabel("Organización política")
plt.ylabel("Votos")
plt.xticks(rotation=80)
plt.tight_layout()
plt.show()
```

```python
blank_votes: int = int(df["VOTOS EN BLANCO"].sum())
null_votes: int = int(df["VOTOS NULOS"].sum())

print("Votos en blanco:", blank_votes)
print("Votos nulos:", null_votes)
```

```python
df["TOTAL_VOTOS_EMITIDOS"] = df[vote_cols].sum(axis=1)

df[["NÚMERO DE MESA", "ELECTORES HÁBILES", "TOTAL_VOTOS_EMITIDOS"]].head()
```

```python
df["PARTICIPACION_ESTIMADA"] = df["TOTAL_VOTOS_EMITIDOS"] / df["ELECTORES HÁBILES"]

df["PARTICIPACION_ESTIMADA"].describe()
```

```python
plt.figure(figsize=(8, 4))
df["PARTICIPACION_ESTIMADA"].hist(bins=30)
plt.title("Distribución de participación estimada por mesa")
plt.xlabel("Participación estimada")
plt.ylabel("Cantidad de mesas")
plt.tight_layout()
plt.show()
```

```python
invalid_votes_df: pd.DataFrame = df[
    df["TOTAL_VOTOS_EMITIDOS"] > df["ELECTORES HÁBILES"]
]

invalid_votes_df[
    [
        "REGIÓN / CONTINENTE",
        "PROVINCIA / PAÍS",
        "DISTRITO / CIUDAD",
        "NÚMERO DE MESA",
        "ELECTORES HÁBILES",
        "TOTAL_VOTOS_EMITIDOS",
    ]
].head()
```

```python
valid_party_cols: List[str] = [
    col for col in vote_cols if col not in ["VOTOS EN BLANCO", "VOTOS NULOS"]
]

total_valid_votes_by_party: pd.Series = (
    df[valid_party_cols].sum().sort_values(ascending=False)
)

total_valid_votes_by_party.head(10)
```

```python
region_party_votes: pd.DataFrame = df.groupby("REGIÓN / CONTINENTE")[
    valid_party_cols
].sum()

winner_by_region: pd.DataFrame = pd.DataFrame(
    {
        "ganador": region_party_votes.idxmax(axis=1),
        "votos": region_party_votes.max(axis=1),
    }
)

winner_by_region.head(20)
```

```python
long_votes_df: pd.DataFrame = df.melt(
    id_vars=[
        "TIPO DE ELECCIÓN",
        "ÁMBITO",
        "REGIÓN / CONTINENTE",
        "PROVINCIA / PAÍS",
        "DISTRITO / CIUDAD",
        "LOCAL DE VOTACIÓN",
        "NÚMERO DE MESA",
        "ESTADO DEL ACTA",
        "ELECTORES HÁBILES",
    ],
    value_vars=vote_cols,
    var_name="ORGANIZACION_POLITICA",
    value_name="VOTOS",
)

long_votes_df.head()
```

```python
processed_dir: Path = Path("../data/official_onpe/processed")
processed_dir.mkdir(parents=True, exist_ok=True)

output_long_path: Path = processed_dir / "resultados_presidencial_long.tsv"

# long_votes_df.to_csv(output_long_path, index=False)
long_votes_df.to_csv(output_long_path, index=False, sep="\t")

output_long_path
```
